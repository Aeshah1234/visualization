```python
from PIL import Image
from io import BytesIO
import requests

url = "https://upload.wikimedia.org/wikipedia/commons/9/9d/Calineczka_VP_ubt.jpeg"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*, */*;q=0.8",
}

response=requests.get(url, headers=headers)
image_file=BytesIO(response.content)
image=Image.open(image_file)
image
```

```python
df = pd.read_csv("movie_quotes.csv")
print(df)
```

```python
from wordcloud import WordCloud
# join all txt from each row from csv
text = " ".join(each for each in df.quote)

# generate wordcloud image 
wordcloud = WordCloud(background_color="white", 
                     colormap = 'inferno').generate(text)

# use matplotlib syntax to put our image in figure 
fig, ax = plt.subplots(figsize=(7,3))
ax.imshow(wordcloud,
          interpolation='bilinear')
ax.axis("off")
```

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])

diagram = venn2_unweighted([A, B],
                            set_labels = ('Fruits', 'Pies'),
                            set_colors=("blue", "red"),
                            alpha =0.5)
plt.show()
```

```python
from matplotlib_venn import venn2_unweighted
import matplotlib.pyplot as plt

A = {"apple", "banana", "watermelon"}
B = {"pumpkin", "blueberry", "apple", "key lime"}

diagram = venn2_unweighted(
    [A, B],
    set_labels=("Fruits", "Pies"),
    set_colors=("blue", "red"),
    alpha=0.5
)
diagram.get_label_by_id("10").set_text("\n".join(A - B))
diagram.get_label_by_id("11").set_text("\n".join(A & B))
diagram.get_label_by_id("01").set_text("\n".join(B - A))

plt.show()
```
